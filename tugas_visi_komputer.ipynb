## 1. Persiapan Citra Input (Auto-Download)
Jalankan cell di bawah ini. Jika file citra belum ada di Google Colab, file akan otomatis diunduh dari repository GitHub sehingga tidak perlu upload manual.

```python
import os
import urllib.request
import cv2
import numpy as np
import matplotlib.pyplot as plt

IMAGE_NAME = '2022-Ferrari-Daytona-SP3-009-1600.jpg'

# Ganti URL berikut dengan link Raw GitHub repository Anda setelah di-push:
# Format: https://raw.githubusercontent.com/<username>/<repo>/main/2022-Ferrari-Daytona-SP3-009-1600.jpg
GITHUB_RAW_URL = 'https://raw.githubusercontent.com/USERNAME/REPO/main/2022-Ferrari-Daytona-SP3-009-1600.jpg'

if not os.path.exists(IMAGE_NAME):
    if 'USERNAME/REPO' not in GITHUB_RAW_URL:
        print(f'Mengunduh citra {IMAGE_NAME} dari GitHub...')
        urllib.request.urlretrieve(GITHUB_RAW_URL, IMAGE_NAME)
        print('Citra berhasil diunduh ke direktori Colab!')
    else:
        print('URL GitHub belum diset. Membuka dialog upload manual:')
        from google.colab import files
        files.upload()
else:
    print(f'File {IMAGE_NAME} sudah ada dan siap digunakan!')
```

## 2. Soal 1: Pemisahan Kanal Warna (Red, Green, Blue)
Memuat citra berwarna lalu menampilkan 3 sub-plot terpisah untuk kanal **Red**, **Green**, dan **Blue** dalam bentuk grayscale.

```python
# 1. Muat citra berwarna
img = cv2.imread(IMAGE_NAME)

# 2. Konversi BGR (default OpenCV) ke RGB (Matplotlib)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 3. Pisahkan kanal warna
R = img_rgb[:, :, 0]  # Kanal Red
G = img_rgb[:, :, 1]  # Kanal Green
B = img_rgb[:, :, 2]  # Kanal Blue

# 4. Tampilkan 3 sub-plot
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(R, cmap='gray')
axes[0].set_title('Kanal Red', fontsize=12)
axes[0].axis('off')

axes[1].imshow(G, cmap='gray')
axes[1].set_title('Kanal Green', fontsize=12)
axes[1].axis('off')

axes[2].imshow(B, cmap='gray')
axes[2].set_title('Kanal Blue', fontsize=12)
axes[2].axis('off')

plt.suptitle('Pemisahan Kanal Warna (RGB) dalam Grayscale', fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig('output_kanal_warna.png', dpi=150, bbox_inches='tight')
plt.show()
print('Grafik berhasil disimpan ke output_kanal_warna.png')
```

## 3. Soal 2: Plot Histogram Intensitas Grayscale
Mengonversi citra ke grayscale dan membuat plot histogram distribusi intensitas kecerahan piksel (rentang 0–255).

```python
# 1. Konversi ke Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 2. Hitung histogram intensitas
histogram = cv2.calcHist([gray], [0], None, [256], [0, 256])

# 3. Plot histogram
plt.figure(figsize=(10, 5))
plt.plot(histogram, color='black', linewidth=1)
plt.fill_between(range(256), histogram.flatten(), alpha=0.3, color='steelblue')
plt.title('Histogram Intensitas Grayscale', fontsize=14)
plt.xlabel('Nilai Intensitas Piksel (0–255)', fontsize=11)
plt.ylabel('Jumlah Piksel', fontsize=11)
plt.xlim([0, 255])
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('output_histogram.png', dpi=150, bbox_inches='tight')
plt.show()
print('Histogram berhasil disimpan ke output_histogram.png')
```
